```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#read csv into dataframes
offers_df = pd.read_csv('https://raw.githubusercontent.com/Walter-Mokwena/Data-analysis/main/data_offers.csv')
orders_df = pd.read_csv('https://raw.githubusercontent.com/Walter-Mokwena/Data-analysis/main/data_orders.csv')
```

```python
print("Offers DataFrame:")
display(offers_df.head())

print("\nOrders DataFrame:")
display(orders_df.head())
```

```python
missing_percentages = (orders_df.isnull().sum() / len(orders_df)) * 100
print("Percentage of missing values in orders_df:")
display(missing_percentages)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def categorize_failure(row):
    # order_status_key: 4 is client cancellation, 9 is rejection/system cancellation
    if row['order_status_key'] == 4:
        if row['is_driver_assigned_key'] == 1:
            return 'Cancelled after driver assignment'
        else:
            return 'Cancelled before driver assignment'
    elif row['order_status_key'] == 9:
        return 'Rejected / No driver found'
    else:
        return 'Successful / Other'

# Apply categorization to orders_df
orders_df['failure_reason'] = orders_df.apply(categorize_failure, axis=1)

# Count the occurrences of each category (excluding Successful / Other to focus on failures)
failure_counts = orders_df[orders_df['failure_reason'] != 'Successful / Other']['failure_reason'].value_value = orders_df['failure_reason'].value_counts()

# Calculate counts and percentages
failure_summary = pd.DataFrame({
    'Count': failure_counts,
    'Percentage': (failure_counts / len(orders_df)) * 100
})

print("Failure Reason Distribution:")
display(failure_summary)

# Visualize the distribution
plt.figure(figsize=(10, 6))
sns.barplot(x=failure_summary.index, y='Count', data=failure_summary, palette='viridis', hue=failure_summary.index, legend=False)
plt.title('Distribution of Orders According to Failure Reasons')
plt.xlabel('Failure Reason')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure order_datetime is treated as a string or datetime to extract the hour
orders_df['hour'] = pd.to_datetime(orders_df['order_datetime'], format='%H:%M:%S', errors='coerce').dt.hour

# Filter out successful orders to focus solely on failures
failed_orders = orders_df[orders_df['failure_reason'] != 'Successful / Other']

# Group by hour and failure reason
hourly_failures = failed_orders.groupby(['hour', 'failure_reason']).size().unstack(fill_value=0)

# Plot the distribution of failed orders by hour
plt.figure(figsize=(14, 7))
hourly_failures.plot(kind='bar', stacked=True, figsize=(14, 7), colormap='viridis', width=0.8)
plt.title('Distribution of Failed Orders by Hour and Reason')
plt.xlabel('Hour of Day')
plt.ylabel('Number of Failed Orders')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Failure Reason')
plt.tight_layout()
plt.show()
```

```python
# Calculate hourly proportions to see if certain hours have an abnormally high proportion of one category
hourly_proportions = hourly_failures.div(hourly_failures.sum(axis=1), axis=0) * 100

plt.figure(figsize=(14, 7))
hourly_proportions.plot(kind='bar', stacked=True, figsize=(14, 7), colormap='coolwarm', width=0.8)
plt.title('Proportional Share of Failure Reasons by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(title='Failure Reason', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Filter down to client cancellations (where cancellation time in seconds is available)
cancelled_orders = orders_df[
    (orders_df['order_status_key'] == 4) &
    (orders_df['cancellations_time_in_seconds'].notnull())
].copy()

# Map driver assignment status
cancelled_orders['driver_status'] = cancelled_orders['is_driver_assigned_key'].map({
    0: 'Without Driver Assigned',
    1: 'With Driver Assigned'
})

# Outlier Removal using the 1.5 * IQR rule on cancellation times
Q1 = cancelled_orders['cancellations_time_in_seconds'].quantile(0.25)
Q3 = cancelled_orders['cancellations_time_in_seconds'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Filter out outliers
cleaned_cancellations = cancelled_orders[
    (cancelled_orders['cancellations_time_in_seconds'] >= lower_bound) &
    (cancelled_orders['cancellations_time_in_seconds'] <= upper_bound)
]

# Group by hour and driver status to find the average cancellation time
hourly_avg_cancel_time = cleaned_cancellations.groupby(['hour', 'driver_status'])['cancellations_time_in_seconds'].mean().unstack()

# Plot the hourly average cancellation time
plt.figure(figsize=(14, 7))
plt.plot(hourly_avg_cancel_time.index, hourly_avg_cancel_time['Without Driver Assigned'], marker='o', label='Without Driver Assigned', color='tomato', linewidth=2)
plt.plot(hourly_avg_cancel_time.index, hourly_avg_cancel_time['With Driver Assigned'], marker='s', label='With Driver Assigned', color='royalblue', linewidth=2)

plt.title('Average Time to Cancellation by Hour (Outliers Removed)', fontsize=14)
plt.xlabel('Hour of Day', fontsize=12)
plt.ylabel('Average Cancellation Time (Seconds)', fontsize=12)
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Driver Assignment Status', fontsize=11)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Group by hour and calculate the average ETA
hourly_avg_eta = orders_df.groupby('hour')['m_order_eta'].mean().reset_index()

# Plotting the average ETA by hour
plt.figure(figsize=(12, 6))
sns.lineplot(x='hour', y='m_order_eta', data=hourly_avg_eta, marker='o', color='forestgreen', linewidth=2.5)
plt.title('Average Estimated Time of Arrival (ETA) by Hour of Day', fontsize=14)
plt.xlabel('Hour of Day', fontsize=12)
plt.ylabel('Average ETA (Seconds)', fontsize=12)
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import h3
import folium
from branca.colormap import LinearColormap

# Map latitude and longitude to H3 resolution 8 hexes using h3 v4 API (latlng_to_cell)
orders_df['h3_hex_8'] = orders_df.apply(
    lambda row: h3.latlng_to_cell(row['origin_latitude'], row['origin_longitude'], res=8),
    axis=1
)

# 1. Calculate how many size 8 hexes contain 80% of all orders
hex_order_counts = orders_df['h3_hex_8'].value_counts().reset_index()
hex_order_counts.columns = ['h3_hex_8', 'order_count']

# Calculate cumulative sum and percentage
hex_order_counts['cumulative_orders'] = hex_order_counts['order_count'].cumsum()
total_orders = len(orders_df)
hex_order_counts['cumulative_percentage'] = (hex_order_counts['cumulative_orders'] / total_orders) * 100

# Filter hexes that constitute the top 80%
top_80_hexes_df = hex_order_counts[hex_order_counts['cumulative_percentage'] <= 80]
if len(top_80_hexes_df) == 0:
    # Fallback to include at least the very top one if the first hex already exceeds 80%
    top_80_hexes_df = hex_order_counts.head(1)
else:
    # Include the border hex that crosses the 80% threshold
    border_index = len(top_80_hexes_df)
    top_80_hexes_df = hex_order_counts.iloc[:border_index + 1]

num_hexes_80 = len(top_80_hexes_df)
print(f"Minimum number of resolution 8 H3 hexes containing 80% of all orders: {num_hexes_80}")
```

```python
# 2. Calculate the number of failures per hex
failed_orders_subset = orders_df[orders_df['failure_reason'] != 'Successful / Other']
hex_failure_counts = failed_orders_subset['h3_hex_8'].value_counts().reset_index()
hex_failure_counts.columns = ['h3_hex_8', 'failure_count']

# Merge failure counts with the top 80% hexes
visualization_df = pd.merge(top_80_hexes_df, hex_failure_counts, on='h3_hex_8', how='left').fillna(0)

# Determine base map center using the mean coordinates of these top hexes
mean_lat = orders_df['origin_latitude'].mean()
mean_lon = orders_df['origin_longitude'].mean()
m = folium.Map(location=[mean_lat, mean_lon], zoom_start=12, tiles='cartodbpositron')

# Color mapping based on failures
max_failures = visualization_df['failure_count'].max()
colormap = LinearColormap(colors=['#fef0d9', '#fdcc8a', '#fc8d59', '#d7301f'], vmin=0, vmax=max_failures)
colormap.add_to(m)

# Draw H3 hexes on the map using h3 v4 API (cell_to_boundary)
for _, row in visualization_df.iterrows():
    hex_id = row['h3_hex_8']
    fail_cnt = int(row['failure_count'])
    order_cnt = int(row['order_count'])

    # Convert H3 to GeoJSON boundary coordinates using h3 v4 API
    hex_boundary = h3.cell_to_boundary(hex_id)

    folium.Polygon(
        locations=hex_boundary,
        fill=True,
        fill_color=colormap(fail_cnt),
        color='gray',
        weight=1,
        fill_opacity=0.6,
        popup=f"Hex ID: {hex_id}<br>Total Orders: {order_cnt}<br>Total Fails: {fail_cnt}"
    ).add_to(m)

display(m)
```
